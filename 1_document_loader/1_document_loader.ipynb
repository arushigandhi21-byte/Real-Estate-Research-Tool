### UnstructuredURLLoader

```python
from pathlib import Path
from langchain_community.document_loaders import BSHTMLLoader, UnstructuredURLLoader

url = "https://www.cnbc.com/2024/12/21/how-the-federal-reserves-rate-policy-affects-mortgages.html"
try:
    docs = UnstructuredURLLoader(urls=[url], continue_on_failure=False).load()
except Exception as exc:
    print(f"Could not load the web page: {exc}")
    docs = []

if not docs or not docs[0].page_content.strip():
    sample_path = Path("sample_article.html")
    if not sample_path.exists():
        sample_path = Path("1_document_loader/sample_article.html")
    docs = BSHTMLLoader(sample_path).load()
    print(f"Using local sample: {sample_path}")

len(docs)
```
